Vector slicing

```python
import numpy as np
```

```python
arr = np.arange(11, 21)
arr
```

```python
arr[9]
```

```python
arr[0:5]
```

```python
arr[:5]
```

```python
arr[3:]
```

```python
arr_slice = arr[0::2]
arr_slice
```

```python
arr1 = np.arange(1,31).reshape(6,5)
arr1
```

```python
arr1[0]
```

```python
arr1[0,0]
```

```python
arr1[5,4]
```

```python
slice1 = arr1[0:2, 1:3]
slice1
```
